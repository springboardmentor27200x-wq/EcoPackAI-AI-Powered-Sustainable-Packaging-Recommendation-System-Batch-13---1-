```python
# =========================================
# Module 4 – AI Recommendation Model
# For Excel file (material.xlsx)
# =========================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

from xgboost import XGBRegressor


# ✅ Read Excel file (CHANGE ONLY HERE)
df = pd.read_excel("material.xlsx")

print(df.head())


# Features
X = df[['strength','weight','biodegradability','recyclability']]

# Targets
y_cost = df['cost']
y_co2 = df['co2']


# Train/Test split
X_train, X_test, y_cost_train, y_cost_test = train_test_split(
    X, y_cost, test_size=0.2, random_state=42
)

_, _, y_co2_train, y_co2_test = train_test_split(
    X, y_co2, test_size=0.2, random_state=42
)


# =========================================
# COST – Random Forest
# =========================================
rf = RandomForestRegressor()
rf.fit(X_train, y_cost_train)

cost_pred = rf.predict(X_test)


# =========================================
# CO2 – XGBoost
# =========================================
xgb = XGBRegressor()
xgb.fit(X_train, y_co2_train)

co2_pred = xgb.predict(X_test)


# =========================================
# Evaluation
# =========================================
print("\nCOST MODEL")
print("RMSE:", np.sqrt(mean_squared_error(y_cost_test, cost_pred)))
print("MAE :", mean_absolute_error(y_cost_test, cost_pred))
print("R2  :", r2_score(y_cost_test, cost_pred))

print("\nCO2 MODEL")
print("RMSE:", np.sqrt(mean_squared_error(y_co2_test, co2_pred)))
print("MAE :", mean_absolute_error(y_co2_test, co2_pred))
print("R2  :", r2_score(y_co2_test, co2_pred))


# =========================================
# Ranking
# =========================================
df['pred_cost'] = rf.predict(X)
df['pred_co2'] = xgb.predict(X)

df['score'] = df['pred_cost'] + df['pred_co2']

ranked_df = df.sort_values('score')

print("\nRecommended Ranking:")
print(ranked_df[['material','pred_cost','pred_co2','score']])


# Save result
ranked_df.to_excel("recommended_materials.xlsx", index=False)
print("\nFile saved: recommended_materials.xlsx")
```
